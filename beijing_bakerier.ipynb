```python
import pandas as pd
import geopandas as gpd

from shapely.geometry import Point
from prcoords import gcj_wgs
from ipyleaflet import Map, GeoData, basemaps, LayersControl
```

```python
gdf = gpd.read_file('data/beijing_bakerier.geojson')
```

```python
gdf['new_latlon'] = gdf['geometry'].apply(lambda x: gcj_wgs((x.y, x.x)))
gdf['geometry'] = gdf['new_latlon'].apply(lambda x: Point(x.lon,x.lat))
```

```python
m = Map(
    basemap=basemaps.OpenStreetMap.Mapnik,
    center=(39.96123, 116.36739 ),
    zoom=10
)

geo_data = GeoData(geo_dataframe = gdf,
    name = 'bakeries')

m.add(geo_data)
m.add(LayersControl())

m
```
