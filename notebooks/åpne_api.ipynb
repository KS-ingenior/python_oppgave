## Bruk av åpne API
Denne notebooken demonstrerer uttrekk fra noen norske åpne geodata APIer. Vi bruker biblioteket [requests](https://pypi.org/project/requests/) til det meste, men hvis ytelse er viktig bør man bruke [AIOHTTP](https://docs.aiohttp.org/en/stable/) istedet.

```python
import json
import requests

import pandas as pd
import geopandas as gpd
import pydeck as pdk

from esridump.dumper import EsriDumper
from shapely.geometry import Point
from ipyleaflet import Map, GeoData, basemaps, LayersControl
```

### NVE
NVE har en rekke [apier](https://api.nve.no/doc/). Vi ser på magasinstatistikk i dette eksempelet.

```python
# Først må vi hente ut polygoner som definerer ELSPOT områder i Norge
# Her kan vi bruke esridump biblioteket for å hente features fra mapserver
d = EsriDumper('https://nve.geodataonline.no/arcgis/rest/services/Mapservices/Elspot/MapServer/0')

elspot_gdf = gpd.GeoDataFrame.from_features(d)

# NVE sitt Magasinstatistikk API kan f.eks gi offentlig data for siste uke for alle ELSPOT områder
response = requests.get(
    'https://biapi.nve.no/magasinstatistikk/api/Magasinstatistikk/HentOffentligDataSisteUke',
    headers={'accept': 'application/json'},
)
ms_df = pd.DataFrame(response.json())

# Endepunktet gir data for vassdragsområder og Norge i helhet, vi er kun interessert i ELSPOT omr.
ms_df = ms_df.loc[ms_df['omrType']=='EL']
# Vi kan nå joine elspot polygoner med statistikk sånn at det kan plottes i kart:
ms_gdf = gpd.GeoDataFrame(elspot_gdf.merge(ms_df, left_on='OBJECTID', right_on='omrnr'))
```

```python
# Vi kan fargelegge kartet utifra endring i fyllingsgrad på denne måten
min_fg = ms_gdf['fyllingsgrad'].min()
max_fg= ms_gdf['fyllingsgrad'].max()
diff = max_fg - min_fg

color_scheme = f"""[
    255 - (255-189) * (fyllingsgrad - {min_fg})/{diff}, 
    255 - 255 * (fyllingsgrad - {min_fg})/{diff}, 
    178 - (178-38) * (fyllingsgrad - {min_fg})/{diff}
]"""

layers = [
    pdk.Layer(
        'GeoJsonLayer',
        ms_gdf,
        opacity=0.3,
        stroked=False,
        filled=True,
        wireframe=True,
        get_fill_color=color_scheme,
        get_line_color=[255, 255, 255],
        pickable=True
    )
]

view_state = pdk.ViewState(
    longitude = 12.307778, 
    latitude = 63.990556,
    zoom = 3
)

r = pdk.Deck(
    layers= layers,
    initial_view_state = view_state,
)
r
```

### Statens vegvesen
Statens vegvesen har gjennom [nvdb](https://nvdbapiles-v3.atlas.vegvesen.no/dokumentasjon) et veldig stort datagrunnlag som kan hentes helt åpent. I dette eksempelet henter vi ut geometri for alle bruer i Norge som er lengre enn 600 meter. Koden må endres hvis du vil hente paginerte resultater (flere enn 500). For bedre ytelse bør også async requests vurderes med AIOHTTP.

```python
response = requests.get('https://nvdbapiles-v3.atlas.vegvesen.no/vegobjekter/60?egenskap="egenskap(1313)>=600"')

# Vi henter først alle bruer over 255 meter
bru_df = pd.DataFrame(response.json()['objekter'])

# Vi spør etter all informasjon for hver bru, denne kan ta lang tid siden vi ikke gjør spørringer asynkront.
bru_df['data'] = bru_df['href'].apply(lambda x: requests.get(x).json())
```

```python
# Vi henter ut alle bruoverflate geometrier som finnes for forskjellige bruer. Ikke alle bruer har overflategeometri fra dette APIet
bru_new_df = pd.DataFrame()
new_index = 0
for _, row in bru_df.iterrows():
    for prop in row['data']['egenskaper']:
        # Egenskap id 8906
        if prop['id'] == 8906:
            new_row = pd.DataFrame({
                'id': row['id'],
                'geometry': prop['verdi']
            }, index=[new_index])
            new_index += 1
            bru_new_df = pd.concat([bru_new_df, new_row])

bru_gdf = gpd.GeoDataFrame(
    bru_new_df,
    geometry=gpd.GeoSeries.from_wkt(bru_new_df['geometry'], crs='EPSG:5973')
).to_crs('EPSG:4326')
```

```python
layers = [
    pdk.Layer(
        'GeoJsonLayer',
        bru_gdf,
        opacity=1,
        stroked=False,
        filled=True,
        wireframe=True,
        get_fill_color=[255,255,255],
        get_line_color=[255, 255, 255],
        pickable=True
    )
]

# Vi viser en spesifikk bru i 3d
view_state = pdk.ViewState( 
    longitude = 5.29599   , 
    latitude = 59.37556 ,
    bearing=20,
    pitch=60,
    zoom = 15
)

r = pdk.Deck(
    layers= layers,
    initial_view_state = view_state,
)
r
```

### met.no
Meterologisk institutt har [åpne API](https://api.met.no/) som lett kan kalles på fra Python. Eksempelet under ser på tjenesten [locationforecast](https://api.met.no/weatherapi/locationforecast/2.0/documentation), som gir oss værmeldingen for en spesifikk lokasjon. Det oppfordres til å identifisere seg for å bruke APIet, les [TOS](https://api.met.no/doc/TermsOfService). Se også [StackOverflow](https://stackoverflow.com/questions/56925702/can-i-execute-a-function-in-apply-to-pandas-dataframe-asynchronously) for å bedre ytelsen.

```python
def get_location_forecast(geometry: Point, method='compact') -> dict:
    """
    Get location forecast for a shapely Point.
    Args:
        geometry (shapely.Point): The point to fetch forecast for.
        method (str): Method to use. One of ['compact', 'complete']
    Returns:
        dict: A dictionary holding the forecast.
    """
    allowed_methods = ['compact', 'complete']
    if method not in allowed_methods:
        raise ValueError(f"method must be one of {allowed_methods}")
    else:
        url = f"https://api.met.no/weatherapi/locationforecast/2.0/{method}"

    headers = {
        'Accept': 'application/json',
        # Identifier to use, change this 
        'User-Agent': 'Læring og testing'
    }
    
    params = {
        'lat': geometry.y,
        'lon': geometry.x,
    }

    response = requests.get(url, headers=headers, params=params)
    forecast = response.json()
    return forecast
```

```python
# Vi leser inn alle store byer i verden, bytt ut med lokasjoner du vil ha værmelding for
cities = gpd.read_file(gpd.datasets.get_path('naturalearth_cities'))

# For hver by kan vi f.eks hente lufttemperatur akkurat nå
# Denne spør met.no om været en by om gangen, så tar ca. 1 minutt.
# Hvis bedre ytelse er nødvendig, bør man identifisere seg bedre / bruke andre API / optimalisere (se link over)
cities['current_air_temperature'] = cities['geometry'].apply(
    lambda x: get_location_forecast(x)['properties']['timeseries'][0]['data']['instant']['details']['air_temperature']
)
```

```python
# Vi kan fargelegge kartet utifra lufttempratur på denne måten
min_temp = cities['current_air_temperature'].min()
max_temp = cities['current_air_temperature'].max()
diff = max_temp - min_temp

color_scheme = f"""[
    255 - (255-189) * (current_air_temperature - {min_temp})/{diff}, 
    255 - 255 * (current_air_temperature - {min_temp})/{diff}, 
    178 - (178-38) * (current_air_temperature - {min_temp})/{diff}
]"""

layers = [
    pdk.Layer(
        "ScatterplotLayer",
        data=cities,
        get_position="geometry.coordinates",
        stroked=True,
        filled=True,
        radius_scale=6,
        radius_min_pixels=2,
        radius_max_pixels=100,
        get_fill_color=color_scheme,
        pickable=True

    )
]

view_state = pdk.ViewState(
    longitude = 0,
    latitude = 20,
    zoom = 1
)

r = pdk.Deck(
    layers= layers,
    initial_view_state = view_state,
)
r
```
